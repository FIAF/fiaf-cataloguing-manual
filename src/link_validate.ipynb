```python
# notebook to manage link validation.

import pathlib
import pydash

markdown = ''

markdown_path = pathlib.Path.cwd().parent / 'markdown'
for x in [x for x in markdown_path.rglob('*') if x.suffix == '.md']:
    with open(x) as markdown_text:
        markdown_text = markdown_text.read()
        markdown += markdown_text

labels = list()        
label_text = markdown.split('\n')
for x in label_text:
    if '\\label' not in x:
        continue
    labels.append(x.split('sec:')[1].replace('}', '').strip())

for l in labels:
    if labels.count(l) > 1:
        raise Exception(l, 'used multiple times.', labels.count(l))
    
# find internal links

internal_links = list()
for x in label_text:
    if '#manual' not in x:
        continue
    internal_links.append(x)

manual_links = list()
for x in internal_links:
    old_link = x.split('#manual-')[1].split(')')[0].strip()
    manual_links.append(f'(#manual-{old_link})')

manual_links = pydash.uniq(manual_links)

print(len(manual_links))

for x in sorted(manual_links):
    print(x)
```
