```python
# notebook to manage link validation.

import pathlib

markdown = ''

markdown_path = pathlib.Path.cwd().parent / 'markdown'
for x in [x for x in markdown_path.rglob('*') if x.suffix == '.md']:
    with open(x) as markdown_text:
        markdown_text = markdown_text.read()
        markdown += markdown_text

labels = list()        
label_text = markdown.split('\n')
for x in label_text:
    if '\\label' not in x:
        continue
    labels.append(x.split('sec:')[1].replace('}', '').strip())

for l in labels:
    if labels.count(l) > 1:
        raise Exception(l, 'used multiple times.', labels.count(l))
```

```python



old_path = pathlib.Path.home() / 'Desktop' / 'old-manual' / 'markdown'
old_manual = [{'filepath':str(x), 'link':x.parent.name.split('-')[1].split('_')[0].upper()} for x in old_path.rglob('*') if x.suffix == '.md']
# print(len(old_manual))

# for o in (old_manual):
#     print(o)

# find internal links

internal_links = list()
for x in label_text:
    if '#manual' not in x:
        continue
    internal_links.append(x)

rep = list()

for x in internal_links:
    old_link = x.split('#manual-')[1].split(')')[0].strip()
    match = [x for x in old_manual if x['link'] == old_link]
    if len(match) != 1:
        continue

    print(match)

    with open(match[0]['filepath']) as old:
        old = old.read().split('\n')
    old = [x for x in old if x != '']
    old = [x for x in old if x[0] != '#']

    ref_new = markdown.split(old[0])

    if len(ref_new) != 2:
        continue

    yes = ref_new[0]
    new_link = yes.split('sec:')[-1].split('}')[0].strip()

    new_link = '\\nameref{sec:'+new_link+'}'


    old_link = x.split('#manual-')[1].split(')')[0].strip()

    old_link = '(#manual-'+old_link+')'

    # print(old_link)
    # print(new_link)

    rep.append({old_link:new_link})

    # replacement = 

#  ([3.1.5.12 Frame Rate](#manual-3.1.5.12)), these values should be the same.


# If the Transfer Speed is the same as the Frame Rate \nameref{sec:frame_rate}

    # print('\n')

print(len(rep))

import json

with open(pathlib.Path.cwd() / 'rep.json', 'w') as out:
    json.dump(rep, out)
```
